```python
# Question 1
# Is there a significant difference in median value of houses bounded by the Charles river or not? 
# (T-test for independent samples)

ax = sns.distplot(boston_df['MEDV'],
                  bins=20,
                  kde=True,
                  color='red',
                  hist_kws={"linewidth": 15,'alpha':1})
ax.set(xlabel='Normal Distribution', ylabel='Frequency')
ax.set_title('Distribution for MEDV')
```

```python
# we can assume it is normal

# Null-Hypothesis H_0 = No mean differences in variance
# Alternative Hypothesis H_1: There are mean differences in variances

#we can use the Levene's Test to check test significance

scipy.stats.levene(boston_df[boston_df['CHAS'] == 1]['MEDV'],
                   boston_df[boston_df['CHAS'] == 0]['MEDV'], center='mean')
```

```python
# Since the p-value is smaller than 0.05 we reject the H0 and can assume differences in variance

# Null-Hypothesis H_0 = No mean value differences between house locations
# Alternative Hypothesis H_1: There are mean value differences between house locations

# T-Test
scipy.stats.ttest_ind(boston_df[boston_df['CHAS'] == 1]['MEDV'],
                   boston_df[boston_df['CHAS'] == 0]['MEDV'], equal_var = False)
```

```python
# Since the p-value is smaller than 0.05 we reject the H0 and can assume there are differences in house values related to the location
```

```python
# Question 2
# Is there a difference in Median values of houses (MEDV) for each proportion of owner occupied units built prior to 1940 (AGE)? 
# (ANOVA)

# Variance test

# Null-Hypothesis H_0 = No differences in mean between 3 groupd mean1 = mean2 = mean3
# Alternative Hypothesis H_1: There are differences in at ,least 1 mean value


scipy.stats.levene(boston_df[boston_df['AGE_group'] == '35 years and younger']['MEDV'],
                   boston_df[boston_df['AGE_group'] == 'between 35 and 70 years']['MEDV'], 
                   boston_df[boston_df['AGE_group'] == '70 years and older']['MEDV'], 
                   center='mean')

```

```python
# Since the p-value is higher than 0.05, the hypothesis cannot be rejected and the variances are equal
```

```python
# Create three new variables

ThirtyFive_lower = boston_df[boston_df['AGE_group'] == '35 years and younger']['MEDV']
ThirtyFive_seventy = boston_df[boston_df['AGE_group'] == 'between 35 and 70 years']['MEDV']
Seventy_older = boston_df[boston_df['AGE_group'] == '70 years and older']['MEDV']
```

```python
# Null-Hypothesis H_0 = No median differences of house values between age groups 
# Alternative Hypothesis H_1: There are mean value differences of house values between age groups

# Run one-way ANOVA

f_statistic, p_value = scipy.stats.f_oneway(ThirtyFive_lower, ThirtyFive_seventy, Seventy_older)
print("F_Statistic: {0}, P-Value: {1}".format(f_statistic,p_value))
```

```python
# Since the p-value is smaller than 0.05, the null hypothesis needs to be rejected and hence there are different means for different age groups
```

```python
# Question 3
# Can we conclude that there is no relationship between Nitric oxide concentrations and proportion of non-retail business acres per town? (Pearson Correlation)

# Null-Hypothesis H_0 = No NOX correlation with INDUS 
# Alternative Hypothesis H_1: There is a NOX correlation with INDUS 

pearson,p_value = scipy.stats.pearsonr(boston_df["NOX"],boston_df["INDUS"])

print("Pearson Coefficient value={0}, P-value={1}".format(pearson,p_value))
```

```python
# Since the p-value is smaller than 0.05, we reject the null hypothesis and therefore we have a correlation between NOX and INDUS
```
